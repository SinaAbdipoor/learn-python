# Data Structures

A variable usually holds one value. A **data structure** holds many values together.
Python has four built-in ones:

| Type | Example | Ordered | Changeable | Duplicates allowed |
|---|---|---|---|---|
| `list` | `[3, 1, 3]` | yes | yes | yes |
| `tuple` | `(3, 1, 3)` | yes | no | yes |
| `dict` | `{"a": 1, "b": 2}` | yes | yes | keys: no |
| `set` | `{3, 1}` | no | yes | no |

## Lists

A `list` is an ordered collection in square brackets `[]`. It's the one you'll use most.

````python
demands = [120, 80, 150, 95]
cities = ["Berlin", "Paris", "Madrid", "Rome"]
mixed = [1, "two", 3.0, True]    # a list can mix types, but usually holds one kind
print(demands)
print(len(demands))
````

### Indexing

Each item has a position called an **index**. Counting starts at **0**, not 1.
Negative indexes count from the end.

| Index | `0` | `1` | `2` | `3` |
|---|---|---|---|---|
| Item | `"Berlin"` | `"Paris"` | `"Madrid"` | `"Rome"` |
| Negative index | `-4` | `-3` | `-2` | `-1` |

````python
print(cities[0])     # first
print(cities[2])     # third
print(cities[-1])    # last
````

````python
# An index that doesn't exist raises an IndexError
try:
    cities[4]
except IndexError as error:
    print("IndexError:", error)
````

### Slicing

`list[start:stop]` gives the items from `start` up to, **but not including**, `stop`.

````python
print(cities[1:3])   # items 1 and 2
print(cities[:2])    # from the start up to 2
print(cities[2:])    # from 2 to the end
print(cities[::2])   # every second item
print(cities[::-1])  # reversed
````

### Changing a list

Lists are **mutable**: you can change them after creating them.

````python
cities[1] = "Lyon"          # replace an item
cities.append("Vienna")     # add to the end
cities.insert(0, "Lisbon")  # add at a position
print(cities)
````

````python
cities.remove("Madrid")     # remove by value
last = cities.pop()         # remove and return the last item
print(last)
print(cities)
````

### Useful list tools

````python
print(sum(demands))
print(min(demands), max(demands))
print(sorted(demands))                 # returns a new sorted list
print(sorted(demands, reverse=True))
print(demands)                         # the original is unchanged
````

````python
demands.sort()    # sorts the list itself (in place) and returns None
print(demands)
````

````python
# Check if something is in a list with `in`
print(150 in demands)
print("Paris" in cities)
````

### Looping over a list

````python
for city in cities:
    print(city)
````

````python
# enumerate() gives you the index and the item
for i, city in enumerate(cities):
    print(i, city)
````

### List comprehensions

A short way to build a new list from an existing one:

```python
new_list = [expression for item in old_list if condition]
```

````python
costs = [10, 25, 7, 40]
with_tax = [c * 1.2 for c in costs]
expensive = [c for c in costs if c > 15]
print(with_tax)
print(expensive)
````

### Watch out: copying a list

`b = a` does **not** copy a list. Both names point to the *same* list, so changing one changes the other.
Use `.copy()` to get a real copy.

````python
a = [1, 2, 3]
b = a          # same list, two names
b.append(4)
print(a)       # a changed too!

c = a.copy()   # a real copy
c.append(5)
print(a)
print(c)
````

## Tuples

A `tuple` is like a list, but **immutable**: once created, it can't be changed.
Use parentheses `()`. Tuples are good for values that belong together and shouldn't change,
like coordinates.

````python
warehouse = (52.52, 13.40)    # (latitude, longitude)
print(warehouse[0])
print(len(warehouse))
````

````python
try:
    warehouse[0] = 48.85
except TypeError as error:
    print("TypeError:", error)
````

````python
# Unpacking: assign each item to its own variable
lat, lon = warehouse
print(lat, lon)
````

````python
# A tuple with one item needs a trailing comma
single = (5,)
not_a_tuple = (5)
print(type(single), type(not_a_tuple))
````

## Dictionaries

A `dict` stores **key: value** pairs in curly braces `{}`.
You look values up by their key instead of by position, like words in a real dictionary.

````python
stock = {"bolts": 500, "nuts": 320, "screws": 150}
print(stock["nuts"])
````

````python
stock["washers"] = 200    # add a new key
stock["bolts"] = 450      # update an existing key
del stock["screws"]       # remove a key
print(stock)
````

````python
# A missing key raises a KeyError...
try:
    stock["gears"]
except KeyError as error:
    print("KeyError:", error)

# ...so use .get() to return a default instead
print(stock.get("gears", 0))
````

````python
print("nuts" in stock)    # `in` checks the keys
print(stock.keys())
print(stock.values())
print(stock.items())
````

### Looping over a dictionary

````python
for item, quantity in stock.items():
    print(f"{item}: {quantity}")
````

````python
# Dictionary comprehension
low_stock = {item: qty for item, qty in stock.items() if qty < 300}
low_stock
````

Keys must be immutable (strings, numbers, tuples), but values can be anything, including lists or other dicts.

````python
# Tuple keys are handy for things like distances between two places
distance = {("Berlin", "Paris"): 1054, ("Paris", "Madrid"): 1271}
print(distance[("Berlin", "Paris")])
````

## Sets

A `set` is an unordered collection of **unique** items in curly braces `{}`.
Duplicates are removed automatically, and there's no indexing.

````python
visited = {"Berlin", "Paris", "Berlin", "Rome"}
print(visited)    # the order may differ, and "Berlin" appears once
print(len(visited))
````

````python
# A quick way to remove duplicates from a list
orders = ["A", "B", "A", "C", "B"]
unique_orders = set(orders)
print(unique_orders)
````

````python
visited.add("Madrid")
visited.discard("Rome")    # no error if it's not there
print(visited)
````

````python
# An empty set is set(), because {} creates an empty dict
print(type({}), type(set()))
````

### Set operations

| Operator | Meaning |
|---|---|
| `a \| b` | union: in `a` or `b` |
| `a & b` | intersection: in both |
| `a - b` | difference: in `a` but not `b` |

````python
supplier_a = {"steel", "copper", "plastic"}
supplier_b = {"copper", "glass", "plastic"}
print(supplier_a | supplier_b)
print(supplier_a & supplier_b)
print(supplier_a - supplier_b)
````

## Nesting

Data structures can hold other data structures. This is how real data usually looks.

````python
products = [
    {"name": "chair", "price": 45.0, "sizes": ["S", "M"]},
    {"name": "table", "price": 120.0, "sizes": ["M", "L"]},
]
print(products[1]["name"])
print(products[0]["sizes"][-1])
````

````python
# A matrix as a list of lists: cost[i][j] is the cost from factory i to store j
cost = [
    [4, 6, 9],
    [5, 3, 8],
]
print(cost[1][2])
````

## Which one should I use?

| Use a... | When you need... | Example |
|---|---|---|
| `list` | an ordered sequence you'll change | daily demands |
| `tuple` | a fixed group of values | a coordinate `(x, y)` |
| `dict` | to look things up by name | stock per item |
| `set` | unique items or fast membership checks | cities already visited |


1. Add a `"ruler"` that costs `2.0`.
2. Create a list `cart` with `"pen"`, `"pen"`, `"notebook"` and `"ruler"`.
3. Calculate the total cost of the cart using `prices`.
