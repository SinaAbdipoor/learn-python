# Variables

A **variable** is a name that refers to a value. You create one with `=` (the *assignment* operator):

```python
name = value
```

Python figures out the type of the value automatically, so you don't need to declare it.

## Numbers

Python has two main number types:
- `int` for whole numbers: `1`, `-7`, `1000`
- `float` for decimal numbers: `3.14`, `-0.5`, `2.0`

````python
x = 1
````

````python
y = 2
````

````python
x + y
````

````python
price = 19.99
quantity = 3
price * quantity
````

### Arithmetic operators

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` | addition | `7 + 2` | `9` |
| `-` | subtraction | `7 - 2` | `5` |
| `*` | multiplication | `7 * 2` | `14` |
| `/` | division (always gives a `float`) | `7 / 2` | `3.5` |
| `//` | floor division (rounds down) | `7 // 2` | `3` |
| `%` | remainder (modulo) | `7 % 2` | `1` |
| `**` | power | `7 ** 2` | `49` |

````python
a = 7
b = 2
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
````

> **Watch out:** decimal numbers are stored approximately, so tiny rounding errors can appear.
> Use `round()` when you display results.

````python
0.1 + 0.2
````

````python
round(0.1 + 0.2, 2)
````

Underscores make large numbers easier to read. Python ignores them:

````python
budget = 1_000_000
budget
````

## Strings

A `str` (string) is text. Wrap it in single `'...'` or double `"..."` quotes, both work the same.

````python
first_name = "Ada"
last_name = 'Lovelace'
first_name + " " + last_name
````

````python
# Strings can be repeated with *
"ha" * 3
````

````python
# len() gives the number of characters
len(first_name)
````

### f-strings

The easiest way to put variables inside text is an **f-string**: put `f` before the quotes and the variable in `{}`.

````python
age = 36
print(f"{first_name} is {age} years old.")
````

````python
# You can format numbers too: .2f means 2 decimal places
total = price * quantity
print(f"Total cost: ${total:.2f}")
````

## Booleans

A `bool` is either `True` or `False` (capital first letter). You usually get one from a comparison.

| Operator | Meaning |
|---|---|
| `==` | equal to |
| `!=` | not equal to |
| `<`, `>` | less than, greater than |
| `<=`, `>=` | less/greater than or equal to |

````python
is_feasible = True
print(is_feasible)
print(5 > 3)
print(5 == 3)
````

````python
# Combine conditions with and, or, not
in_stock = True
on_sale = False
print(in_stock and on_sale)
print(in_stock or on_sale)
print(not on_sale)
````

> **Watch out:** `=` assigns a value, `==` compares two values.

## None

`None` means "no value yet". It's useful as a placeholder.

````python
best_solution = None
print(best_solution)
````

## Checking the type

Use `type()` to see what type a value has.

````python
print(type(42))
print(type(3.14))
print(type("hello"))
print(type(True))
print(type(None))
````

## Converting between types

Use `int()`, `float()`, `str()` and `bool()` to convert values.

````python
print(int("42") + 1)    # string -> int
print(float("2.5"))     # string -> float
print(str(100) + "%")   # int -> string
print(int(9.99))        # float -> int cuts off the decimals (doesn't round)
````

You can't add a string and a number directly. Python raises a `TypeError`:

````python
try:
    "Age: " + 36
except TypeError as error:
    print("TypeError:", error)
````

````python
# Fix: convert the number, or use an f-string
print("Age: " + str(36))
print(f"Age: {36}")
````

## Reassigning variables

A variable can be changed at any time, even to a different type.

````python
counter = 0
counter = counter + 1
counter += 1    # shorthand for counter = counter + 1
counter
````

The shorthand works for other operators too: `-=`, `*=`, `/=`.

## Multiple assignment

Assign several variables in one line, or swap two values without a temporary variable.

````python
width, height = 4, 5
print(width, height)

width, height = height, width
print(width, height)
````

## Naming rules

**Rules (Python enforces these):**
- Use only letters, digits and underscores: `total_cost`, `x1`
- Don't start with a digit: `1st_value` is an error
- No spaces or hyphens: `total cost` and `total-cost` are errors
- Names are case-sensitive: `cost` and `Cost` are different variables
- Don't use Python keywords like `if`, `for`, `class`, `True`

**Conventions (style, from PEP 8):**
- Use `snake_case` for variables: `max_capacity`, not `maxCapacity`
- Pick descriptive names: `unit_cost` is clearer than `uc`
- Use `UPPER_CASE` for values that shouldn't change (constants)

````python
MAX_CAPACITY = 500
unit_cost = 12.5
units_produced = 40

total_cost = unit_cost * units_produced
print(f"Total cost: {total_cost}, capacity used: {units_produced / MAX_CAPACITY:.0%}")
````
